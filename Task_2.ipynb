```python
import requests
import pandas as pd
from frost_id import client_id
import matplotlib.pyplot as plt

client_id = client_id

#Requests the daily mean air temperture for all of 2025
endpoint = 'https://frost.met.no/observations/v0.jsonld'
parameters = {
'sources': 'SN17850',
'elements': 'mean(air_temperature P1D)',"sum(precipitation_amount P1D)"
'referencetime': '2025-01-01/2025-12-31',
}

# Issue an HTTP GET request
r = requests.get(endpoint, parameters, auth=(client_id,''))
# Extract JSON data
data = r.json()



# Hvis 'df' har kolonnene 'tid', 'temp_min' og 'temp_maks':
data.plot(x="tid", y=["Precipitation", "temperature"], kind="line")

plt.title("Rain and temperature over time in 2025")
plt.ylabel("Temperature (°C)")
plt.grid(True)
plt.show()
```
